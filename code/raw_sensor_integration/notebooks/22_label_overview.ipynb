# Label Overview

Renders 3-panel PNGs for every echogram: the Sv echogram, the bottom-detection mask, and the mask overlaid in red on the echogram. Output is organised by frequency into subfolders. Both zarr stores and raw files are supported.

```python
import sys

sys.path.insert(0, '../..')
from project_paths import REPO_ROOT

BASE = REPO_ROOT

ZARR_DIR = BASE / 'Datasets' / 'echopype_zarr'
RAW_DIRS = [
    BASE / 'Datasets' / 'raw_files' / 'drive-download-20260317T113707Z-1-001',
]
OUTPUT_DIR = BASE / 'label_overview'
```

## Generate PNGs

Run the standalone script. Add `--raw-dir` arguments to include raw files.

```python
raw_args = ' '.join(f'--raw-dir "{d}"' for d in RAW_DIRS)
cmd = f'python ../../visualize_labels.py --zarr-dir "{ZARR_DIR}" {raw_args} --output-dir "{OUTPUT_DIR}"'
print(cmd)
!{cmd}
```

## Output summary

```python
for sub in sorted(OUTPUT_DIR.iterdir()):
    if not sub.is_dir():
        continue
    print(f"\n{sub.name}/")
    for freq_dir in sorted(sub.iterdir()):
        if not freq_dir.is_dir():
            continue
        pngs = list(freq_dir.glob('*.png'))
        print(f"  {freq_dir.name}/  ({len(pngs)} PNGs)")
```

## Sample images

```python
from IPython.display import display, Image

sample_pngs = []
for sub in sorted(OUTPUT_DIR.iterdir()):
    if not sub.is_dir():
        continue
    for freq_dir in sorted(sub.iterdir()):
        if not freq_dir.is_dir():
            continue
        pngs = sorted(freq_dir.glob('*.png'))
        if pngs:
            sample_pngs.append(pngs[0])

for png in sample_pngs[:6]:
    print(f"\n{png.relative_to(OUTPUT_DIR)}")
    display(Image(filename=str(png), width=900))
```
